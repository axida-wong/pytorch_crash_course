张量的结构类似于 [Numpy](https://numpy.org/)中的ndarrays，而张量可以运行在GPU及其他相似的硬件加速器上。 事实上，为了减少数据的拷贝，张量和NumPy arrays在底层常常共享同一块内存(`bridge-to-np-label`{.interpreted-text role=”ref”})。 在自动微分(automatic differentiation)的过程中也使用张量进行优化(在后续[Autograd](autogradqs_tutorial.html)章节可以看到更多有关内容)。 如果已经对ndarrays十分熟悉了，那对张量的API也可以运用自如。如果还不熟悉，下面的教程会帮助你上手。

```python
import torch
import numpy as np
```

```python
# 初始化张量, 会自动推断类型

data = [[1,2], [3, 4]]
x_data = torch.tensor(data)
x_data
```

```python
# 使用numpy创建

np_array = np.array(data)
x_np = torch.from_numpy(np_array)
x_np
```

```python
# 使用已有张量创建

x_ones = torch.ones_like(x_data)
print(f"Ones Tensor:\n {x_ones}\n")

x_zeros = torch.zeros_like(x_data)
print(f"Ones Tensor:\n {x_zeros}\n")

x_rand = torch.rand_like(x_data, dtype=torch.float)
print(f"Random Tensor\n {x_rand}\n")
```

```python
# 通过随机或者常量创建

shape = (2, 3)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print("Random Tensor")
print(rand_tensor)
print("Ones Tensor")
print(ones_tensor)
print("Zeros Tensor")
print(zeros_tensor)
```

```python
# 张量的属性保存了其形状，数据类型，以及其存储设备类型。

tensor = torch.rand(3, 4)

print(f"shape of tensor           : {tensor.shape}")
print(f"datatype of tensor        : {tensor.dtype}")
print(f"device tensor is stored on: {tensor.device}")
tensor = tensor.to("cuda")
print(f"\nnow the tensor is moved to cuda.")
print(f"device tensor is stored on: {tensor.device}")
```

```python
# 张量操作

t = torch.arange(1000).reshape(50,-1)
print(t.shape)
if torch.cuda.is_available():
    t = t.to("cuda")

print(t.device)
```

```python
tensor = torch.ones(4, 4)

print(f"First row: {tensor[0]}")
print(f"First col: {tensor[:, 0]}")
print(f"last col: {tensor[:, -1]}")

tensor[:, 0] = 0

print(tensor)
```

```python
# concat

t1 = torch.cat([tensor, tensor, tensor], dim=1)
print(t1)
```

```python
# 运筹操作

tensor = torch.randint(4, (2,2))

y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)

y3 = torch.ones_like(y1)
torch.matmul(tensor, tensor.T, out=y3)

print(tensor, y1, y2, y3, sep='\n')
```

```python
# 单个元素的张量 在聚合运算场景中，你可能会得到一个单元素的张量，
# 可使用`item()`将其传唤为Python数值。

agg = tensor.sum()
print(agg, type(agg))
agg_item = agg.item()
print(agg_item, type(agg_item))
```

```python
# 原地操作 修改张量中的原值操作称为原地操作。它们以 _ 后缀表示。
# 例如：x.copy_(y)，x.t_()，会改变 x。

tensor = torch.arange(12).reshape(3, 4)
print(tensor)

tensor.add_(5)
print(tensor)

t2 = torch.randint(1, 12, (3, 4))
print(t2)

tensor.copy_(t2)
print(tensor, t2, sep='\n')
print(id(tensor), id(t2))

```

```python
# 张量在使用CPU时，可与NumPy arrays共享内存空间，修改其中一个
# 会同步映射到另一个上。

t = torch.ones(5)
print(f"t: {t}")
n = t.numpy()
print(f"n: {n}")
```

```python
# 对于张量的修改体现到了NumPy array上。

t.add_(1)
print(f"t: {t}")
print(f"n: {n}")
```

```python
# Numpy array 转换为张量

n = np.ones(5)
t = torch.from_numpy(n)

print(n, type(n))
print(t, type(t))

np.add(n, 1, out=n)
print(n, type(n))
print(t, type(t))
```
